# Retrieval baseline
Evaluate the current local production retriever without modifying its artifacts.

```python
from pathlib import Path

from app.api.services.retriever_loader import RetrieverLoader
from research.src.datasets import load_jsonl, split_queries
from research.src.resource_metrics import build_manifest, save_manifest, set_seed
from research.src.runner import evaluate_retriever

SEED = 42
ARTIFACTS_DIR = Path('../../artifacts/artifacts_rag_baseline_latest')
set_seed(SEED)
records = load_jsonl(Path('../datasets/retrieval_smoke.jsonl'))
documents, queries = split_queries(records)
```

```python
retriever = RetrieverLoader(str(ARTIFACTS_DIR)).load()
metrics, rows = evaluate_retriever(retriever, queries, k=5)
index_size = sum(path.stat().st_size for path in ARTIFACTS_DIR.rglob('*') if path.is_file())
manifest = build_manifest(
    experiment='retrieval-baseline',
    seed=SEED,
    models=['current-production-retriever'],
    metrics=metrics,
    peak_rss_bytes=0,
    index_size_bytes=index_size,
)
save_manifest('../results/00_retrieval_baseline.json', manifest)
metrics
```
