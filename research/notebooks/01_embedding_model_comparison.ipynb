# Embedding model comparison
Measure quality, latency, RAM, and index size before choosing a model. Models are loaded one at a time.

```python
RUN_HEAVY_MODELS = False
SEED = 42
MODEL_IDS = ['sentence-transformers/all-mpnet-base-v2', 'BAAI/bge-m3']
```

```python
if RUN_HEAVY_MODELS:
    import gc
    import torch
    from sentence_transformers import SentenceTransformer

    device = 'mps' if torch.backends.mps.is_available() else 'cpu'
    measurements = []
    for model_id in MODEL_IDS:
        model = SentenceTransformer(model_id, device=device)
        # Run the shared evaluator and record quality, latency, peak RSS, and index size.
        measurements.append({'model_id': model_id, 'status': 'measurement required'})
        del model
        gc.collect()
else:
    print('Heavy comparison disabled. Set RUN_HEAVY_MODELS=True manually.')
```
